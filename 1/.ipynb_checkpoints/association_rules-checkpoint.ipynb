
# Ассоциативные правила: Apriori и FP-Growth

В ноутбуке рассматриваются два классических алгоритма поиска частых наборов:

1. **Apriori** — только классический вариант с генерацией кандидатов `JOIN` и отсевом `PRUNE`;
2. **FP-Growth** — поиск частых наборов без явной генерации всех кандидатов, с использованием FP-дерева.

После поиска частых наборов для обоих алгоритмов выполняется единый этап:

- генерация ассоциативных правил;
- расчёт метрик;
- фильтрация правил по нескольким критериям.

Используется библиотека `mlxtend`.



## 1. Ассоциативные правила

Ассоциативное правило записывается в виде

$
X \rightarrow Y,
$

где:

- $X$ — antecedent, или условие;
- $Y$ — consequent, или следствие;
- $X \cap Y = \varnothing$.

Например:

$
\{\text{Пиво},\text{Хлеб}\}
\rightarrow
\{\text{Колбаса}\}.
$

Ассоциативное правило показывает статистическую связь между наборами объектов, но само по себе не доказывает причинно-следственную зависимость.

Поиск правил выполняется в два этапа:

$
\boxed{\text{поиск частых наборов}}
$

$
\boxed{\text{генерация и оценка правил}}
$


## 2. Основные метрики

Пусть $N$ — число транзакций.

### Support

$
support(X)=\frac{\sigma(X)}{N},
$

где $|X|$ — число транзакций, содержащих набор X.

Для правила:

$
support(X\rightarrow Y)=support(X\cup Y).
$

### Confidence

$
confidence(X\rightarrow Y)
=
\frac{support(X\cup Y)}{support(X)}.
$

Это условная вероятность появления \(Y\) при наличии \(X\).

### Lift

$
lift(X\rightarrow Y)
=
\frac{confidence(X\rightarrow Y)}{support(Y)}
=
\frac{support(X\cup Y)}
{support(X)\cdot support(Y)}.
$

Интерпретация:

- lift>1 — положительная связь;
- lift=1 — независимость;
- lift<1 — отрицательная связь.

В рассматриваемом учебном примере показатель `impr` соответствует `lift`.

### Leverage

$
leverage(X\rightarrow Y)
=
support(X\cup Y)
-
support(X)\cdot support(Y).
$

### Conviction

$
conviction(X\rightarrow Y)
=
\frac{1-support(Y)}
{1-confidence(X\rightarrow Y)}.
$

При confidence=1:

$
conviction=\infty.
$



## 3. Частые наборы и правила — разные этапы

Очень важно не смешивать поиск частых наборов и оценку правил.

### Поиск частых наборов

Используется условие:

$
support(X)\ge minsup.
$

Именно по `support` работают Apriori и FP-Growth.

### Оценка правил

После получения частых наборов рассчитываются:

- `confidence`;
- `lift`;
- `leverage`;
- `conviction`;
- дополнительные метрики.

Поэтому слабое правило, например

$
\text{Пиво}\rightarrow\text{Хлеб},
$

не означает, что набор

$
\{\text{Пиво},\text{Хлеб}\}
$

надо удалить из множества частых наборов.



# Часть I. Классический Apriori

## 4. Принцип Apriori

Классический Apriori использует свойство антимонотонности поддержки:

$
A\subseteq B
\quad\Rightarrow\quad
support(B)\le support(A).
$

Отсюда следует правило:

> Если хотя бы одно подмножество кандидата не является частым, то сам кандидат также не может быть частым.

Для формирования кандидатов размера k:

1. **JOIN** — объединить подходящие частые наборы из L_{k-1};
2. **PRUNE** — удалить кандидат, если хотя бы одно его подмножество размера k-1 отсутствует в L_{k-1};
3. для оставшихся кандидатов рассчитать `support`;
4. сформировать L_k.


## 5. Псевдокод классического Apriori

```text
Apriori(D, minsup)

C1 ← все одноэлементные кандидаты

для каждого X ∈ C1:
    вычислить support(X)

L1 ← {X ∈ C1 | support(X) ≥ minsup}

k ← 2

пока L(k-1) не пусто:

    Ck ← JOIN(L(k-1))

    для каждого кандидата X ∈ Ck:
        если существует подмножество S ⊂ X,
        |S| = k-1 и S ∉ L(k-1):
            удалить X из Ck              # PRUNE

    для каждого оставшегося X ∈ Ck:
        вычислить support(X)

    Lk ← {X ∈ Ck | support(X) ≥ minsup}

    k ← k + 1

вернуть L1 ∪ L2 ∪ ... ∪ L(k-1)
```

Схема:

```text
Транзакции
    ↓
C1 → support → L1
                 ↓
              JOIN
                 ↓
                C2
                 ↓
              PRUNE
                 ↓
              support
                 ↓
                L2
                 ↓
              JOIN
                 ↓
                C3
                 ↓
               ...
```


````python
import inspect
from itertools import combinations

import numpy as np
import pandas as pd

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 120)
````

````python

# ============================================================
# Пороговые значения
# ============================================================

MIN_SUPPORT = 0.50
MIN_CONFIDENCE = 0.60
MIN_LIFT = 1.00
MIN_CONVICTION = 1.00

MIN_LEVERAGE = 0.00
MIN_ZHANG = 0.00
MIN_CERTAINTY = 0.00

# Учебные пороги для дополнительных метрик
MIN_JACCARD = 0.50
MIN_KULCZYNSKI = 0.50
MIN_REPRESENTATIVITY = 1.00
````


## 6. Исходные транзакции

Используем 6 транзакций:

| № | Товары |
|---|---|
| 1 | Пиво, Колбаса, Подгузники, Хлеб |
| 2 | Молоко, Пиво, Колбаса, Детское питание, Подгузники, Хлеб |
| 3 | Молоко, Кола, Хлеб |
| 4 | Молоко, Пиво, Колбаса, Подгузники, Хлеб |
| 5 | Пиво, Кола, Детское питание |
| 6 | Пиво, Колбаса, Подгузники |

При:

\[
minsup=0.5
\]

минимальная частота равна:

$
6\cdot0.5=3.
$


````python

dataset = [
    ["Пиво", "Колбаса", "Подгузники", "Хлеб"],
    ["Молоко", "Пиво", "Колбаса", "Детское питание", "Подгузники", "Хлеб"],
    ["Молоко", "Кола", "Хлеб"],
    ["Молоко", "Пиво", "Колбаса", "Подгузники", "Хлеб"],
    ["Пиво", "Кола", "Детское питание"],
    ["Пиво", "Колбаса", "Подгузники"],
]

transactions_df = pd.DataFrame({
    "Транзакция": range(1, len(dataset) + 1),
    "Товары": [", ".join(row) for row in dataset],
})

transactions_df
````

````python
# One-hot кодирование
te = TransactionEncoder()
te_array = te.fit(dataset).transform(dataset)

df = pd.DataFrame(te_array, columns=te.columns_)
df
````


## 7. Реализация классического Apriori

Ниже вручную реализованы:

$
C_1\rightarrow L_1,
\quad
C_2\rightarrow L_2,
\quad
C_3\rightarrow L_3,
\quad \ldots
$

При этом для $k\ge3$ обязательно выполняется `PRUNE`.


````python
transaction_sets = [set(t) for t in dataset]

# считает, в скольких транзакциях встречается набор
def support_count(itemset):
    itemset = set(itemset)
    return sum(itemset.issubset(t) for t in transaction_sets)

# рассчитывает поддержку набора
def support_value(itemset):
    return support_count(itemset) / len(transaction_sets)

# преобразует набор товаров в удобную строку
def itemset_str(itemset):
    return "{" + ", ".join(sorted(itemset)) + "}"

# формирует таблицу для кандидатов
def candidates_table(candidates):
    rows = []
    for c in sorted(candidates, key=lambda x: (len(x), sorted(x))):
        supp = support_value(c)
        rows.append({
            "itemset": itemset_str(c),
            "freq": support_count(c),
            "support": supp,
            "частый": supp >= MIN_SUPPORT,
        })
    return pd.DataFrame(rows)

# реализует основной шаг классического Apriori
def apriori_join_prune(prev_frequent, k):
    prev_frequent = set(prev_frequent)
    candidates = set()

    # JOIN
    for a, b in combinations(prev_frequent, 2):
        candidate = a | b

        if len(candidate) != k:
            continue

        # PRUNE
        all_subsets_frequent = all(
            frozenset(subset) in prev_frequent
            for subset in combinations(candidate, k - 1)
        )

        if all_subsets_frequent:
            candidates.add(frozenset(candidate))

    return candidates

# оставляет только те кандидаты, для которых support≥MIN_SUPPORT
def frequent_from_candidates(candidates):
    return {
        c for c in candidates
        if support_value(c) >= MIN_SUPPORT
    }
````

````python

# ============================================================
# C1 → L1
# ============================================================

all_items = sorted(set().union(*transaction_sets))

C1 = {frozenset([item]) for item in all_items}
L1 = frequent_from_candidates(C1)

print("C1")
display(candidates_table(C1))

print("L1")
display(candidates_table(L1))
````

````python

# ============================================================
# C2 → L2
# ============================================================

items_L1 = sorted(next(iter(s)) for s in L1)

C2 = {
    frozenset(c)
    for c in combinations(items_L1, 2)
}
L2 = frequent_from_candidates(C2)

print("C2")
display(candidates_table(C2))

print("L2")
display(candidates_table(L2))
````

````python

# ============================================================
# C3 → L3
# ============================================================

C3 = apriori_join_prune(L2, 3)
L3 = frequent_from_candidates(C3)

print("C3 после JOIN + PRUNE")
display(candidates_table(C3))

print("L3")
display(candidates_table(L3))
````


### Почему `{Молоко, Пиво, Хлеб}` отсутствует в \(C_3\)

Для кандидата

$
\{\text{Молоко},\text{Пиво},\text{Хлеб}\}
$

необходимо проверить:

$
\{\text{Молоко},\text{Пиво}\},
\quad
\{\text{Молоко},\text{Хлеб}\},
\quad
\{\text{Пиво},\text{Хлеб}\}.
$

Но:

$
support(\{\text{Молоко},\text{Пиво}\})
=
\frac{2}{6}
=
0.333<0.5.
$

Следовательно,

$
\{\text{Молоко},\text{Пиво}\}\notin L_2,
$

и тройка удаляется на этапе `PRUNE` **до подсчёта её поддержки**.

````python

target_triple = frozenset({"Молоко", "Пиво", "Хлеб"})

check_rows = []
for s in combinations(target_triple, 2):
    s = frozenset(s)
    check_rows.append({
        "Подмножество": itemset_str(s),
        "freq": support_count(s),
        "support": support_value(s),
        "Входит в L2": s in L2,
    })

pd.DataFrame(check_rows)
````

````python

# ============================================================
# C4 → L4
# ============================================================

C4 = apriori_join_prune(L3, 4)
L4 = frequent_from_candidates(C4)

print("C4 после JOIN + PRUNE")
display(candidates_table(C4))

print("L4")
display(candidates_table(L4))
````


## 8. Классический Apriori средствами `mlxtend`

Теперь тот же поиск выполняется встроенной функцией:

```python
mlxtend.frequent_patterns.apriori
```

Результат должен совпасть с ручной классической реализацией.


````python

frequent_itemsets_apriori = apriori(
    df,
    min_support=MIN_SUPPORT, 
    use_colnames=True #в результате используются названия товаров
).copy()
# добавляет размер набора
frequent_itemsets_apriori["length"] = frequent_itemsets_apriori["itemsets"].apply(len)
# переводит относительную поддержку в абсолютную частоту
frequent_itemsets_apriori["freq"] = (
    frequent_itemsets_apriori["support"] * len(df)
).round().astype(int)
#преобразует frozenset в удобную строку
frequent_itemsets_apriori["itemsets_text"] = (
    frequent_itemsets_apriori["itemsets"].apply(itemset_str)
)
# итоговая таблица только с нужными столбцами
frequent_itemsets_apriori = frequent_itemsets_apriori.sort_values(
    ["length", "support", "itemsets_text"],
    ascending=[True, False, True]
).reset_index(drop=True)

frequent_itemsets_apriori[
    ["length", "itemsets_text", "freq", "support"]
]
````

````python
for k in sorted(frequent_itemsets_apriori["length"].unique()):
    print(f"L{k}")
    display(
        frequent_itemsets_apriori.loc[
            frequent_itemsets_apriori["length"] == k,
            ["itemsets_text", "freq", "support"]
        ].reset_index(drop=True)
    )
````


# Часть II. FP-Growth

## 9. Идея FP-Growth

**FP-Growth (Frequent Pattern Growth)** решает ту же задачу поиска частых наборов, но не создаёт $C_2,C_3,C_4,\ldots$ явным перебором кандидатов.

Основная идея:

1. один проход по базе — определить частоты товаров;
2. удалить товары с `support < minsup`;
3. отсортировать частые товары по убыванию частоты;
4. второй проход по базе — построить **FP-tree**;
5. для каждого товара построить условную базу шаблонов;
6. сформировать условное FP-дерево;
7. рекурсивно извлечь частые наборы.

FP-Growth обычно эффективнее Apriori на больших и плотных наборах транзакций, потому что:

- не генерирует огромное количество кандидатов;
- сжимает повторяющиеся префиксы транзакций в дереве;
- требует меньше повторных проходов по исходной базе.



## 10. Схема FP-Growth

```text
Исходные транзакции
        ↓
Подсчитать частоты товаров
        ↓
Удалить support < minsup
        ↓
Отсортировать товары по частоте
        ↓
Построить FP-tree
        ↓
Для каждого частого товара:
        ↓
Conditional Pattern Base
        ↓
Conditional FP-tree
        ↓
Рекурсивный поиск
        ↓
Частые наборы
```

В отличие от Apriori:

```text
Apriori:
L1 → C2 → L2 → C3 → L3 → C4 → ...

FP-Growth:
транзакции → FP-tree → условные FP-tree → частые наборы
```



## 11. Псевдокод FP-Growth

```text
FP_Growth(D, minsup)

1. Подсчитать support всех элементов.

2. Удалить элементы:
       support < minsup

3. Отсортировать оставшиеся элементы
   по убыванию частоты.

4. Построить FP-tree:
       для каждой транзакции:
           удалить нечастые элементы
           отсортировать оставшиеся элементы
           добавить путь в FP-tree
           увеличить счётчики общих узлов

5. Для каждого элемента i,
   начиная с наименее частого:

       сформировать conditional pattern base(i)

       построить conditional FP-tree(i)

       если conditional FP-tree(i) не пусто:
           рекурсивно найти частые шаблоны

6. Вернуть все найденные частые наборы.
```


````python

# ============================================================
# FP-Growth средствами mlxtend
# ============================================================

frequent_itemsets_fp = fpgrowth(
    df,
    min_support=MIN_SUPPORT,
    use_colnames=True
).copy()

frequent_itemsets_fp["length"] = frequent_itemsets_fp["itemsets"].apply(len)
frequent_itemsets_fp["freq"] = (
    frequent_itemsets_fp["support"] * len(df)
).round().astype(int)
frequent_itemsets_fp["itemsets_text"] = (
    frequent_itemsets_fp["itemsets"].apply(itemset_str)
)

frequent_itemsets_fp = frequent_itemsets_fp.sort_values(
    ["length", "support", "itemsets_text"],
    ascending=[True, False, True]
).reset_index(drop=True)

frequent_itemsets_fp[
    ["length", "itemsets_text", "freq", "support"]
]
````


## 12. Сравнение результатов Apriori и FP-Growth

При одинаковом `min_support` алгоритмы должны найти **одинаковое множество частых наборов**.

Различается только способ поиска.


````python

def normalize_itemsets(frame):
    return {
        frozenset(row.itemsets): float(row.support)
        for row in frame[["itemsets", "support"]].itertuples(index=False)
    }

apriori_result = normalize_itemsets(frequent_itemsets_apriori)
fp_result = normalize_itemsets(frequent_itemsets_fp)

same_itemsets = set(apriori_result) == set(fp_result)
same_supports = all(
    np.isclose(apriori_result[itemset], fp_result[itemset])
    for itemset in apriori_result.keys() & fp_result.keys()
)

comparison_algorithms = pd.DataFrame([
    {
        "Алгоритм": "Apriori",
        "Количество частых наборов": len(frequent_itemsets_apriori),
        "Генерирует кандидаты Ck": "Да",
        "Основная структура": "Ck и Lk",
    },
    {
        "Алгоритм": "FP-Growth",
        "Количество частых наборов": len(frequent_itemsets_fp),
        "Генерирует кандидаты Ck": "Нет",
        "Основная структура": "FP-tree",
    },
])

display(comparison_algorithms)

print("Одинаковые наборы:", same_itemsets)
print("Одинаковые support:", same_supports)
````


## 13. Apriori и FP-Growth: основные различия

| Характеристика | Apriori | FP-Growth |
|---|---|---|
| Генерация кандидатов | Да | Нет |
| Структура | \(C_k\), \(L_k\) | FP-tree |
| Основная операция | JOIN + PRUNE | построение и рекурсивный анализ дерева |
| Проходы по базе | Многократные | обычно 2 основных прохода для построения дерева |
| Память | может расти из-за числа кандидатов | требуется хранить FP-tree |
| Плотные данные | может быть медленным | обычно эффективнее |
| Учебная наглядность | очень высокая | сложнее |
| Итог | частые наборы | те же частые наборы |

Для маленького учебного набора разница по времени практически несущественна. Преимущество FP-Growth проявляется на больших данных.



# Часть III. Генерация ассоциативных правил

## 14. Из какого результата строить правила?

Поскольку Apriori и FP-Growth при одинаковом `min_support` дают одинаковые частые наборы, правила также будут одинаковыми.

Ниже для определённости используется результат Apriori:

```python
frequent_itemsets_apriori
```

Его можно заменить на:

```python
frequent_itemsets_fp
```

без изменения логики дальнейшего расчёта.


````python
#задаёт параметры генерации правил
association_kwargs = {
    "metric": "support",
    "min_threshold": 0.0,
}

# Совместимость с версиями mlxtend, где требуется num_itemsets
if "num_itemsets" in inspect.signature(association_rules).parameters:
    association_kwargs["num_itemsets"] = len(df)
# строит ассоциативные правила из частых наборов
rules = association_rules(
    frequent_itemsets_apriori[["support", "itemsets"]],
    **association_kwargs
).copy()

rules["impr"] = rules["lift"]
# преобразует левую часть правила в удобную строку
rules["antecedents_text"] = rules["antecedents"].apply(itemset_str)
# аналогично оформляет правую часть
rules["consequents_text"] = rules["consequents"].apply(itemset_str)
#список основных метрик, которые всегда выводятся
base_columns = [
    "antecedents_text",
    "consequents_text",
    "support",
    "confidence",
    "lift",
    "impr",
    "leverage",
    "conviction",
]
# добавляет дополнительные метрики
extra_columns = [
    c for c in [
        "representativity",
        "zhangs_metric",
        "jaccard",
        "certainty",
        "kulczynski",
    ]
    if c in rules.columns
]
# Последний блок сортирует правила
rules[
    base_columns + extra_columns
].sort_values(
    ["lift", "confidence"],
    ascending=False
).reset_index(drop=True)
````


## 15. Базовый фильтр правил

Для демонстрации используем:

$
support\ge0.5,
$

$
confidence\ge0.6,
$

$
lift>1,
$

$
conviction>1.
$

`lift > 1` и `conviction > 1` выбирают положительную ассоциацию.


````python

rules_basic = rules[
    (rules["support"] >= MIN_SUPPORT)
    & (rules["confidence"] >= MIN_CONFIDENCE)
    & (rules["lift"] > MIN_LIFT)
    & (rules["conviction"] > MIN_CONVICTION)
].copy()

rules_basic[
    [
        "antecedents_text",
        "consequents_text",
        "support",
        "confidence",
        "lift",
        "impr",
        "leverage",
        "conviction",
    ]
].sort_values(
    ["lift", "confidence"],
    ascending=False
).reset_index(drop=True)
````


## 16. Отсев сразу по всем доступным метрикам

Дополнительные пороги ниже являются учебными. Для реальной задачи они должны подбираться отдельно.


````python

FILTERS = {
    "support": (">=", MIN_SUPPORT),
    "confidence": (">=", MIN_CONFIDENCE),
    "lift": (">", MIN_LIFT),
    "leverage": (">", MIN_LEVERAGE),
    "conviction": (">", MIN_CONVICTION),
    "zhangs_metric": (">", MIN_ZHANG),
    "jaccard": (">=", MIN_JACCARD),
    "certainty": (">", MIN_CERTAINTY),
    "kulczynski": (">=", MIN_KULCZYNSKI),
    "representativity": (">=", MIN_REPRESENTATIVITY),
}

def apply_condition(series, operator, threshold):
    if operator == ">":
        return series > threshold
    if operator == ">=":
        return series >= threshold
    if operator == "<":
        return series < threshold
    if operator == "<=":
        return series <= threshold
    if operator == "==":
        return series == threshold
    raise ValueError(f"Неизвестный оператор: {operator}")

rules_checked = rules.copy()
available_filters = {}

for metric, (operator, threshold) in FILTERS.items():
    if metric not in rules_checked.columns:
        continue

    available_filters[metric] = (operator, threshold)
    rules_checked[f"pass_{metric}"] = apply_condition(
        rules_checked[metric],
        operator,
        threshold,
    )

pass_columns = [
    f"pass_{metric}"
    for metric in available_filters
]

rules_checked["accepted_all_metrics"] = (
    rules_checked[pass_columns].all(axis=1)
)

print("Применённые пороги:")
for metric, (operator, threshold) in available_filters.items():
    print(f"  {metric:18s} {operator} {threshold}")
````

````python

rules_all_metrics = rules_checked[
    rules_checked["accepted_all_metrics"]
].copy()

show_cols = [
    "antecedents_text",
    "consequents_text",
    "support",
    "confidence",
    "lift",
    "leverage",
    "conviction",
]

show_cols += [
    c for c in [
        "zhangs_metric",
        "jaccard",
        "certainty",
        "kulczynski",
        "representativity",
    ]
    if c in rules_all_metrics.columns
]

rules_all_metrics[
    show_cols
].sort_values(
    ["lift", "confidence"],
    ascending=False
).reset_index(drop=True)
````


## 17. Пример: `Пиво → Хлеб`

$
support=0.5,
$

$
confidence=0.6,
$

$
lift=0.9,
$

$
conviction\approx0.833.
$

Набор:

$
\{\text{Пиво},\text{Хлеб}\}
$

остаётся частым, потому что его `support = 0.5`.

Но само правило:

$
\text{Пиво}\rightarrow\text{Хлеб}
$

можно отбросить по условиям:

$
lift>1,
\qquad
conviction>1.
$


````python

beer_to_bread = rules_checked[
    (rules_checked["antecedents"] == frozenset({"Пиво"}))
    & (rules_checked["consequents"] == frozenset({"Хлеб"}))
]

diagnostic_columns = [
    "antecedents_text",
    "consequents_text",
    "support",
    "confidence",
    "lift",
    "leverage",
    "conviction",
] + pass_columns + ["accepted_all_metrics"]

beer_to_bread[
    [c for c in diagnostic_columns if c in beer_to_bread.columns]
]
````


## 18. Сильное правило более высокого порядка

Несмотря на слабое правило

$
\text{Пиво}\rightarrow\text{Хлеб},
$

можно получить сильное правило:

$
\{\text{Пиво},\text{Хлеб}\}
\rightarrow
\{\text{Колбаса}\}.
$

Для него:

$
confidence=1,
$

$
lift=1.5,
$

$
conviction=\infty.
$

````python
beer_bread_to_sausage = rules_checked[
    (rules_checked["antecedents"] == frozenset({"Пиво", "Хлеб"}))
    & (rules_checked["consequents"] == frozenset({"Колбаса"}))
]

beer_bread_to_sausage[
    [c for c in diagnostic_columns if c in beer_bread_to_sausage.columns]
]
````

### Главное

- **Apriori** и **FP-Growth** решают одну задачу — находят частые наборы.
- При одинаковом `min_support` итоговые частые наборы должны совпадать.
- Apriori использует кандидаты и `JOIN + PRUNE`.
- FP-Growth использует FP-дерево и не генерирует \(C_k\).
- Метрики `confidence`, `lift`, `conviction` применяются **после** поиска частых наборов.
